```python
# Step 1: Import required libraries
import pandas as pd

from google.colab import files

# Step 2: Upload the file from your local machine
uploaded = files.upload()

# Step 3: Read the Excel file into a DataFrame
# (Replace 'your_file.xlsx' with your actual filename)
file_name = next(iter(uploaded))  # Gets the first uploaded filename
df = pd.read_excel(file_name)

df
```

```python
# Step 2:
df1 = df.drop_duplicates()
df1
```

```python
# Step 3:
df2 = df1.drop(columns = 'Not_Useful_Column')
df2
```

```python
# Step 4:
df2['Last_Name'] = df2['Last_Name'].str.strip("123._/")
df2
```

```python
# Step 5:

import numpy as np

def clean_phone_number(phone):
    # Check for NaN or None values
    if pd.isna(phone):
        return np.nan

    # Convert to string if it isn't already
    phone = str(phone)

    # Handle 'N/a', 'n/a', 'NA', etc. cases
    if phone.lower() in ['n/a', 'na', 'NaN']:
        return np.nan

    # Remove all non-digit characters
    cleaned = ''.join(filter(str.isdigit, phone))

    # Format as XXX-XXX-XXXX if we have exactly 10 digits
    if len(cleaned) == 10:
        return f"{cleaned[:3]}-{cleaned[3:6]}-{cleaned[6:]}"
    else:
        return "np.nan"  # Return NaN for invalid lengths

# Apply the cleaning function to the Phone_Number column
df2['Phone_Number'] = df2['Phone_Number'].apply(clean_phone_number)

# Optional: Drop rows with NaN phone numbers if you don't want them
df3 = df2.dropna(subset=['Phone_Number'])

# Display the cleaned data
df3
```

```python
# Step 6:
def split_address(address):
    if pd.isna(address) or str(address).lower() == 'n/a':
        return (np.nan, np.nan)

    address = str(address)
    # Split on first space
    parts = address.split(' ', 1)
    if len(parts) == 2:
        return parts[0], parts[1]
    else:
        return (np.nan, address)

# Create final cleaned dataframe (df4) with address split
df4 = df3.copy()
df4[['Street_Number', 'Street_Name']] = df4['Address'].apply(
    lambda x: pd.Series(split_address(x))
)

# Reorder columns
cols = ['CustomerID', 'First_Name', 'Last_Name', 'Phone_Number',
        'Street_Number', 'Street_Name',
        'Paying Customer', 'Do_Not_Contact']
df4 = df4[cols]

# Display the cleaned data
print("Final cleaned dataframe (df4):")
display(df4)

# Verify all variables exist
print("\nAvailable dataframes:")
print(f"Original: df2 (shape: {df2.shape})")
print(f"After phone cleaning: df3 (shape: {df3.shape})")
print(f"Final cleaned: df4 (shape: {df4.shape})")
```

```python
# Step 7:
df4['Paying Customer'] = df4['Paying Customer'].replace({
    'Yes': 'Y',
    'No': 'N'
})

df4['Do_Not_Contact'] = df4['Do_Not_Contact'].replace({
    'Yes': 'Y',
    'No': 'N'
})

df4
```

```python
# Step 8:
df5 = df4.replace('N/a', np.nan)
df5 = df5.fillna('')
df5
```
