**VARIoT-dataset**

Detta dataset innehåller ***35 250 nätverksflöden*** från IoT-enheter (***smarta kameror, röstassistenter*** m.m.) under normal drift.

**Huvudegenskaper:**
Källor: Lokala IoT-enheter (192.168.20.xx) kommunicerar med externa servrar (54.x.x.x) eller broadcast (255.255.255.255).

***Protokoll:*** TCP (port 80/443) och UDP (***ovanliga portar som 6666***).

**Tid**: Flöden varar ***från millisekunder till minuter*** (t.ex. 117 sekunder för UDP-broadcast).

**Användningsområden:**
Obejakad ***anomalidetektering*** (***inga etiketter – bra för Isolation Forest***).

**Identifiera avvikelser:**

***Ovanliga portar*** (***6666*** kan vara skadlig).

***Extremt korta/långa flöden.***

***TCP-flaggor (PSH/URG) som inte är typiska för IoT.***
